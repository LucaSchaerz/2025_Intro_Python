## Exercise 6: Missing data and interpolation

Given `series = np.array([1.0, 2.0, np.nan, 4.0, 5.0])`, fill the missing value by linear interpolation against the integer index, then print both the NaN-aware mean and the ordinary mean to show the difference.

````python
series = np.array([1.0, 2.0, np.nan, 4.0, 5.0])

mask = np.isnan(series)
known = np.where(~mask)[0]      # indices valides   : [0 1 3 4]
missing = np.where(mask)[0]     # indices manquants : [2]

filled = series.copy()
filled[mask] = np.interp(missing, known, series[known])

print(filled)
print("NaN-aware mean :", np.nanmean(series))
print("ordinary mean  :", np.mean(series))
````

$$\rho - \rho_{\text{pure water}} = a\,S + b\,\Theta + c\,\Theta^{2}$$


Source: Roquet et al. (2015), *J. Phys. Oceanogr.* 45, 2564–2579, and the
[TEOS-10 GSW toolbox](https://www.teos-10.org/).
